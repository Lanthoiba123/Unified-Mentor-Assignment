## Numpy copy vs view

```python
# x1=
x1 = np.array([1, 2, 3, 4, 5])
print(x1)
```

```python
# assign x2 = x1
x2 = x1
```

```python
#print x1 and x2
print("x1:", x1)
print("x2:", x2)
```

Ok now you have seen that both of them are same

```python
# change 1st element of x2 as 10
x2[0] = 10
```

```python
#Again print x1 and x2
print("x1:", x1)
print("x2:", x2)
```

```python
# Check memory share between x1 and x2
print(np.shares_memory(x1, x2))
```

Hey It's True they both share memory

Shall we try **view()** function also likwise.

```python
# Create a view of x1 and store it in x3.
x3 = x1.view()
```

```python
# Again check memory share between x1 and x3
print(np.shares_memory(x1, x3))
```

Woh! simple assignment is similar to view. That means
The view does not own the data and any changes made to the view will affect the original array, and any changes made to the original array will affect the view.

Don't agree? ok lets change x3 and see if original array i.e. x1 also changes

```python
#Change 1st element of x3=100
x3[0] = 100
```

```python
#print x1 and x3 to check if changes reflected in both
print("x1:", x1)
print("x3:", x3)
```

Now its proved.

Lets see how **Copy()** function works

```python
# Now create an array x4 which is copy of x1
x4 = x1.copy()
print(x4)
```

```python
# Change the last element of x4 as 900
x4[-1] = 900
```

```python
# print both x1 and x4 to check if changes reflected in both
print("x1:", x1)
print("x4:", x4)
```

Hey! such an intresting output. You noticed buddy! your original array didn't get changed on change of its copy ie. x4.

Still not convinced? Ok lets see if they both share memory or not

```python
#Check memory share between x1 and x4
print(np.shares_memory(x1, x4))
```

**hstack vs vstack function**


Stacking is same as concatenation, the only difference is that stacking is done along a new axis.

NumPy provides a helper function:

1. hstack() to stack along rows.
2. vstack()  to stack along columns

```python
# stack x1 and x4 along columns.
stacked_cols = np.vstack((x1, x4))
print(stacked_cols)
```

```python
#stack x1 and x4 along rows
stacked_rows = np.hstack((x1, x4))
print(stacked_rows)
```

We hope now you saw the difference between them.

Fun fact! you can even use concatenate() function to join 2 arrays along with the axis. If axis is not explicitly passed, it is taken as 0 ie. along column

Lets try this function as well

```python
# arr1 =
arr1 = np.array([1, 2, 3])
# arr2 =
arr2 = np.array([4, 5, 6])

##join arr1 and arr2 along rows using concatenate() function
joined_rows = np.concatenate((arr1, arr2), axis=0)
print(joined_rows)
```

```python
##join arr1 and arr2 along columns using concatenate() function
# Stacking 1D arrays horizontally requires matching dimensions
joined_cols = np.concatenate((arr1.reshape(-1, 1), arr2.reshape(-1, 1)), axis=1)
print(joined_cols)
```
